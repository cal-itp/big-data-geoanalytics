```python
# A function to save credentials after client authentication

saved_credentials = {}

def save_credentials(username, database, server, session_id):
    saved_credentials["username"] = username
    saved_credentials["database"] = database
    saved_credentials["server"] = server
    saved_credentials["session_id"] = session_id

def get_saved_credentitals():
    return saved_credentials
```

```python
#If already authenticated and credentials are saved, use those
#else re-authenticate with the Altitude API
prev_credentials = get_saved_credentitals()

if prev_credentials:
    client = AltitudeAPI(username=prev_credentials["username"], database=prev_credentials["database"], server=prev_credentials["server"], session_id=prev_credentials["session_id"])
else:
    pass
    client = AltitudeAPI(username=account_credentials["username"], password=account_credentials["password"], database=account_credentials["database"], server=account_credentials["server"])
    credentials = client.authenticate()
    save_credentials(username=credentials.username, database=credentials.database, server=credentials.server, session_id=credentials.session_id)
```
